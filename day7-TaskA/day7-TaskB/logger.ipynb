```python
import logging

import cv2 as cv
import numpy as np

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s",
    force=True,
)
logger = logging.getLogger(__name__)


def read_image(path: str) -> np.ndarray:
    """Read an image from disk; raise FileNotFoundError if it can't be read."""
    logger.info("Reading %s", path)
    img = cv.imread(path)
    if img is None:
        logger.error("File not found: %s", path)
        raise FileNotFoundError(f"Could not read image: {path}")
    logger.info("Loaded shape=%s", img.shape)
    return img


for file_name in ("flower.jpg", "wrong.jpg"):
    try:
        read_image(file_name)
    except FileNotFoundError as e:
        print("Error:", e)
```

```python
import logging

import cv2 as cv
import numpy as np

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s",
    handlers=[logging.FileHandler("app.log"), logging.StreamHandler()],
    force=True,
)
logger = logging.getLogger(__name__)


def read_image(path: str) -> np.ndarray:
    """Read an image from disk; raise FileNotFoundError if it can't be read."""
    logger.info("Reading %s", path)
    img = cv.imread(path)
    if img is None:
        logger.error("File not found: %s", path)
        raise FileNotFoundError(f"Could not read image: {path}")
    logger.info("Loaded shape=%s", img.shape)
    return img


for file_name in ("flower.jpg", "wrong.jpg"):
    try:
        read_image(file_name)
    except FileNotFoundError:
        logger.exception("Image load failed")
```
