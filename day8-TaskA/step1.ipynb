```python
import cv2 as cv
import matplotlib.pyplot as plt

img = cv.imread("flower.jpg")

if img is None:
    print("File not found")
else:
    img_gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

    print("gray shape:", img_gray.shape)
    print("gray dtype:", img_gray.dtype)
    print("gray min, max:", img_gray.min(), img_gray.max())

    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title("Colour")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(img_gray, cmap="gray")
    plt.title("Gray")
    plt.axis("off")

    plt.show()
```

```python
img = cv.imread("flower.jpg")

if img is None:
    print("File not found")
else:
    h, w = img.shape[:2]

    half = cv.resize(img, (w // 2, h // 2),interpolation=cv.INTER_NEAREST)
    wrong = cv.resize(img, (h // 2, w // 2))

    print("img shape:", img.shape)
    print("half shape:", half.shape)
    print("wrong shape:", wrong.shape)

    half_rgb = cv.cvtColor(half, cv.COLOR_BGR2RGB)
    plt.imshow(half_rgb)
    plt.title("Half size")
    plt.axis("off")
    plt.show()
    
    wrong_rgb = cv.cvtColor(wrong ,cv.COLOR_BGR2RGB)
    plt.imshow(wrong_rgb)
```
