```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2 as cv
img = cv.imread("flower.jpg")

if img is None:
    print("File not found")
else:
    gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

    negative = 255 - gray
    negative_cv = cv.bitwise_not(gray)

    print("Same as bitwise_not:", np.array_equal(negative, negative_cv))
    print("negative dtype:", negative.dtype)
    print("gray[0, 0]:", gray[0, 0], "| negative[0, 0]:", negative[0, 0])
    print("Sum of the pair:", int(gray[0, 0]) + int(negative[0, 0]))

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(gray, cmap="gray")
    plt.title("Gray")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(negative, cmap="gray")
    plt.title("Negative")
    plt.axis("off")

    plt.show()
```
